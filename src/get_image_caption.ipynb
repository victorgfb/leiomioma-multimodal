```python
base_path = "/run/media/victor/pessoal/mestrado/codigo/datasets/leiomyoma_files/articles"
```

```python
from pathlib import Path

xmls = []
for p in Path(base_path).rglob("*.nxml"):
    xmls.append(str(p))
```

```python
len(xmls)
```

```python
from bs4 import BeautifulSoup
import pandas as pd
import re

art_captions_data = []
pmc_pattern = r'PMC\d+'

for xml in xmls:
    art = re.search(pmc_pattern, xml).group()

    with open(xml, 'r') as f:
        file = f.read()

        soup = BeautifulSoup(file, 'lxml')
        figures = soup.find_all('fig')
        for figure in figures:
            label = figure.find('label')
            caption = figure.find('caption')
            graphic = figure.find('graphic')
            
            if caption is not None and label is not None and graphic is not None:
                caption_text = caption.text.strip()
                fig_id = graphic.get('xlink:href')
                art_captions_data.append([label.text, caption_text, fig_id, art])

art_captions_df = pd.DataFrame(columns=["label", "caption", "id", "article_id"], data=art_captions_data)
```
